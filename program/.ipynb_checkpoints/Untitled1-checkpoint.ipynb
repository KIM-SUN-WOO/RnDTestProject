```python
import pandas as pd

ExampleData='4.xlsx'
df1=pd.DataFrame()
df1=pd.read_excel('../data/'+ExampleData)
df1=df1.dropna()

#df_ant - projection to z axis for anterior view
df_ant = df1[['l_hip_x','l_hip_y','l_knee_x','l_knee_y','l_ankle_x','l_ankle_y',
             'r_hip_x','r_hip_y','r_knee_x','r_knee_y','r_ankle_x','r_ankle_y']]

pd.set_option('mode.chained_assignment', None)
df_ant['l_tibia_x'] = df_ant['l_knee_x'] - df_ant['l_ankle_x']
df_ant['l_tibia_y'] = df_ant['l_knee_y'] - df_ant['l_ankle_y']
df_ant['l_femur_x'] = df_ant['l_hip_x'] - df_ant['l_knee_x']
df_ant['l_femur_y'] = df_ant['l_hip_y'] - df_ant['l_knee_y']
df_ant['r_tibia_x'] = df_ant['r_knee_x'] - df_ant['r_ankle_x']
df_ant['r_tibia_y'] = df_ant['r_knee_y'] - df_ant['r_ankle_y']
df_ant['r_femur_x'] = df_ant['r_hip_x'] - df_ant['r_knee_x']
df_ant['r_femur_y'] = df_ant['r_hip_y'] - df_ant['r_knee_y']

#lateral view
df_lat = df1[['torso_y', 'torso_z', 'waist_y','waist_z','l_hip_y','l_hip_z','l_knee_y','l_knee_z','l_ankle_y','l_ankle_z']]

df_lat['tor_wai_y'] = df_lat['torso_y'] - df_lat['waist_y']
df_lat['tor_wai_z'] = df_lat['torso_z'] - df_lat['waist_z']
df_lat['l_hip_wai_y'] = df_lat['waist_y'] - df_lat['l_hip_y']
df_lat['l_hip_wai_z'] = df_lat['waist_z'] - df_lat['l_hip_z']
df_lat['l_tibia_y'] = df_lat['l_knee_y'] - df_lat['l_ankle_y']
df_lat['l_tibia_z'] = df_lat['l_knee_z'] - df_lat['l_ankle_z']
```

```python
def angle_v2(vec1, vec2):
    #compute angle between 2 vectors. ... from google
    #import math
    import numpy as np

    np.array 
    
    # 각 벡터의 각도 계산 (x축 기준)
    angle1 = np.arctan2(vec1[:,1], vec1[:,0])
    angle2 = np.arctan2(vec2[:,1], vec2[:,0])

    # 두 각도의 차이 계산
    angle_diff_rad = angle2 - angle1

    # 절대값 및 360도 조정 (필요에 따라)
    angle_diff_deg = np.degrees(angle_diff_rad)
    #angle_diff_deg = np.where( angle_diff_deg < 0, angle_diff_deg + 360, angle_diff_deg )

    return angle_diff_deg
```

```python
import matplotlib.pyplot as plt
import numpy as np
import io
from PIL import Image

def get_img_from_fig(fig, dpi=100):
    """
    Convert a matplotlib figure to a 3D numpy array of RGBA pixels.
    
    @param fig a matplotlib figure
    @param dpi the dots per inch resolution
    @return a numpy 3D array of RGBA values
    """
    with io.BytesIO() as buf:
        # Save the figure to the buffer in PNG format
        fig.savefig(buf, format="png", dpi=dpi, bbox_inches='tight', pad_inches=0)
        buf.seek(0)
        # Open the image from the buffer using PIL
        image = Image.open(buf)
        # Convert the PIL image to a NumPy array (H, W, C)
        img_as_np = np.array(image)
    return img_as_np

# Example usage
# 1. Create a figure and plot something
#fig, ax = plt.subplots()
#ax.plot([1, 2, 3], [4, 5, 6])
#ax.set_title("Sample Plot")
#plt.close(fig) # Close the figure to prevent it from being displayed

# 2. Get the figure as a NumPy array
#pixel_array = get_img_from_fig(fig)

#print("Image shape:", pixel_array.shape)
## The shape will typically be (height, width, 4) for RGBA channels
```

```python
%matplotlib widget
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation

import io
def fig2array(fig):
    """Convert a Matplotlib figure to a NumPy array and return it"""
    # Draw the renderer
    fig.canvas.draw()
    # Get the RGBA array of the figure
    img_data = np.frombuffer(fig.canvas.tostring_rgb(), dtype=np.uint8)
    # Get the figure size
    w, h = fig.canvas.get_width_height()
    # Reshape the array to a 3D array with RGB channels
    img_array = img_data.reshape((h, w, 3))
    return img_array
    

plt.clf()
fig, ax = plt.subplots()

images = []
for i in range( 50):
    
    ax.plot( df_ant[['l_ankle_x','l_knee_x','l_hip_x']].loc[i].values,
                df_ant[['l_ankle_y','l_knee_y','l_hip_y']].loc[i].values,"o-r" )

    ax.plot( df_ant[['r_ankle_x','r_knee_x','r_hip_x']].loc[i].values, 
              df_ant[['r_ankle_y','r_knee_y','r_hip_y']].loc[i].values,"o-r" )

    img = get_img_from_fig(fig)
    images.append([img])
    if(i>5):
        break
output = animation.ArtistAnimation(fig, images, interval=50, blit=True)


plt.show()
```

```python
%matplotlib widget
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

# 1. Set up the Figure
fig, ax = plt.subplots(1,2)
#fig.supylabel('Y(vertical)')
fig.suptitle('Anterior view (Torso to Ankle) - '+ExampleData)
plt.subplots_adjust(wspace=0.5)
# Set plot limits
ax[0].set_xlim(-270, 240)
ax[0].set_ylim(-1450, 50)
ax[0].set_xlabel('X(Left-Right)', fontsize=8)
ax[0].set_ylabel('Y(vertical)', fontsize=8)
ax[0].tick_params(axis='both',  labelsize=8)
ax[0].set_title('Original')
    
ax[1].set_xlim(-270, 240)
ax[1].set_ylim(-1450, 50)
ax[1].set_xlabel('X(Left-Right)', fontsize=8)
ax[1].set_ylabel('Y(vertical)', fontsize=8)
ax[1].tick_params(axis='both',  labelsize=8)
ax[1].set_title('Filtered')


left_leg, = ax[0].plot([], [], lw=2, color='red', marker='o',label='left leg')
right_leg, = ax[0].plot([], [], lw=2, color='red',marker='o', label='right leg')
lines = [left_leg, right_leg] # Store lines in a list for easy iteration

left_leg2, = ax[1].plot([], [], lw=2, color='red', marker='o',label='left leg')
right_leg2, = ax[1].plot([], [], lw=2, color='red',marker='o', label='right leg')
lines2 = [left_leg2, right_leg2] # Store lines in a list for easy iteration


frame_text = ax[0].text(0.02, 0.95, '', transform=ax[0].transAxes, fontsize=8)
frame_text2 = ax[1].text(0.02, 0.95, '', transform=ax[1].transAxes, fontsize=8)
# 3. Initialize the Plot (optional but good practice)
def init():
    for i in lines :
        i.set_data([], [])
    for i in lines2 :
        i.set_data([], [])
    frame_text.set_text('')
    frame_text2.set_text('')
    return lines, lines2, frame_text, frame_text2, # Return the artist(s) to be animated

# 4. Create the Update Function
def animate(i):

    # Update data for each frame

    left_leg.set_data( df1[['l_ankle_x','l_knee_x','l_hip_x','waist_x','torso_x']].loc[i].values,
                       df1[['l_ankle_y','l_knee_y','l_hip_y','waist_y','torso_y']].loc[i].values)
    right_leg.set_data(df1[['r_ankle_x','r_knee_x','r_hip_x','waist_x']].loc[i].values, 
                  df1[['r_ankle_y','r_knee_y','r_hip_y','waist_y']].loc[i].values)
    
    left_leg2.set_data( df1_filtered[['l_ankle_x','l_knee_x','l_hip_x','waist_x','torso_x']].loc[i].values,
                       df1_filtered[['l_ankle_y','l_knee_y','l_hip_y','waist_y','torso_y']].loc[i].values)
    right_leg2.set_data(df1_filtered[['r_ankle_x','r_knee_x','r_hip_x','waist_x']].loc[i].values, 
                  df1_filtered[['r_ankle_y','r_knee_y','r_hip_y','waist_y']].loc[i].values)
                       
    frame_text.set_text(f'Frame: {i:03d}') # Formats the number to have leading zeros
    frame_text2.set_text(f'Frame: {i:03d}') # Formats the number to have leading zeros
    return lines, lines2, frame_text, frame_text2 # Return the artist(s) that were modified

# 5. Create the FuncAnimation Object
# interval is the delay between frames in milliseconds (e.g., 20ms = 50 fps)
# blit=True optimizes drawing by only updating changed parts
ani = FuncAnimation(fig, animate, frames=df_ant.shape[0], init_func=init,
                    interval=1, blit=True, repeat=False)

# 6. Show the Animation
plt.show()

# To save the animation as a GIF or MP4, you can use the save method:
ani.save('sine_wave.gif', writer='pillow', fps=28) 
```

```python
%matplotlib widget
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

# 1. Set up the Figure
fig, ax = plt.subplots()
xdata, ydata = [], []
line, = ax.plot([], [], lw=2) # Initialize an empty line object

# Set plot limits
ax.set_xlim(0, 2 * np.pi)
ax.set_ylim(-1.1, 1.1)

# 3. Initialize the Plot (optional but good practice)
def init():
    line.set_data([], [])
    return line, # Return the artist(s) to be animated

# 4. Create the Update Function
def animate(i):
    # Update data for each frame
    x = np.linspace(0, 2 * np.pi, 1000)
    y = np.sin(x + 0.01 * i) # Shift the sine wave based on frame number 'i'
    line.set_data(x, y)
    return line, # Return the artist(s) that were modified

# 5. Create the FuncAnimation Object
# interval is the delay between frames in milliseconds (e.g., 20ms = 50 fps)
# blit=True optimizes drawing by only updating changed parts
ani = FuncAnimation(fig, animate, frames=200, init_func=init,
                    interval=20, blit=True)

# 6. Show the Animation
plt.show()

# To save the animation as a GIF or MP4, you can use the save method:
# ani.save('sine_wave.gif', writer='pillow', fps=50) 
```

```python
%matplotlib widget
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation

# 1. Setup the figure and axes
fig, ax = plt.subplots()
ax.set_xlim(0, 2*np.pi)
ax.set_ylim(-1.1, 1.1)

# 2. Create the initial, empty line objects (artists)
# The comma unpacks the single Line2D object returned by ax.plot()
line1, = ax.plot([], [], lw=2, color='blue', label='sin')
line2, = ax.plot([], [], lw=2, color='red', label='cos')
lines = [line1, line2] # Store lines in a list for easy iteration
ax.legend()

# Initialize data storage for the animation
xdata, y1data, y2data = [], [], []

# 3. Define the initialization function (optional but recommended for blitting)
def init():
    """Resets the lines' data at the start of the animation."""
    line1.set_data([], [])
    line2.set_data([], [])
    return lines # Must return an iterable of all artists to be updated

# 4. Define the animation (update) function
def animate(frame):
    """
    Updates the data for each line in the frame.
    The 'frame' argument is automatically passed by FuncAnimation.
    """
    # Append new data for the current frame
    xdata.append(frame)
    y1data.append(np.sin(frame))
    y2data.append(np.cos(frame))

    # Update the lines with the new data
    line1.set_data(xdata, y1data)
    line2.set_data(xdata, y2data)

    return lines # Must return an iterable of all artists to be updated

# 5. Create the FuncAnimation object
# Store the animation object in a variable to prevent it from being garbage collected
ani = animation.FuncAnimation(
    fig,                  # The figure to animate
    animate,              # The function called for each frame
    frames=np.linspace(0, 2*np.pi, 128), # Data source for the 'frame' argument
    init_func=init,       # The initialization function
    interval=20,          # Delay between frames in milliseconds
    blit=True             # Optimize drawing by only updating changed parts
)

# 6. Show the animation
plt.show()
```
